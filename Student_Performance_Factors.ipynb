```python
import matplotlib.pyplot as plt

# 1. Attendance vs Average Exam Score
attendance_groups = pd.cut(
    df["Attendance"],
    bins=[59, 69, 79, 89, 100],
    labels=["60–69%", "70–79%", "80–89%", "90–100%"]
)

attendance_scores = df.groupby(
    attendance_groups, observed=False
)["Exam_Score"].mean()

plt.figure(figsize=(8, 5))
plt.bar(attendance_scores.index, attendance_scores.values)
plt.xlabel("Attendance Range")
plt.ylabel("Average Exam Score")
plt.title("Average Exam Score by Attendance Range")
plt.savefig("Student_Performance_Analysis/attendance_vs_exam_score.png",
            bbox_inches="tight")
plt.show()


# 2. Hours Studied vs Average Exam Score
hours_groups = pd.cut(
    df["Hours_Studied"],
    bins=[0, 10, 20, 30, 40, 50],
    labels=["1–10", "11–20", "21–30", "31–40", "41–50"]
)

hours_scores = df.groupby(
    hours_groups, observed=False
)["Exam_Score"].mean()

plt.figure(figsize=(8, 5))
plt.bar(hours_scores.index, hours_scores.values)
plt.xlabel("Hours Studied")
plt.ylabel("Average Exam Score")
plt.title("Average Exam Score by Hours Studied")
plt.savefig("Student_Performance_Analysis/hours_studied_vs_exam_score.png",
            bbox_inches="tight")
plt.show()


# 3. Parental Involvement
parental_scores = df.groupby("Parental_Involvement")["Exam_Score"].mean()

plt.figure(figsize=(7, 5))
plt.bar(parental_scores.index, parental_scores.values)
plt.xlabel("Parental Involvement")
plt.ylabel("Average Exam Score")
plt.title("Parental Involvement vs Average Exam Score")
plt.savefig("Student_Performance_Analysis/parental_involvement.png",
            bbox_inches="tight")
plt.show()


# 4. Access to Resources
resource_scores = df.groupby("Access_to_Resources")["Exam_Score"].mean()

plt.figure(figsize=(7, 5))
plt.bar(resource_scores.index, resource_scores.values)
plt.xlabel("Access to Resources")
plt.ylabel("Average Exam Score")
plt.title("Access to Resources vs Average Exam Score")
plt.savefig("Student_Performance_Analysis/access_to_resources.png",
            bbox_inches="tight")
plt.show()

print("All 4 charts saved successfully!")
```
